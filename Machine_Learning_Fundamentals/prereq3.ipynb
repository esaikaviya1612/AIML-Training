````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def has_passed(self):
        return self.marks >= 40

    def grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 40:
            return "C"
        else:
            return "Fail"

    def summary(self):
        status = "passed" if self.has_passed() else "failed"
        return f"{self.name} scored {self.marks} and {status} (grade {self.grade()})."

s1 = Student("KK", 88)
s2 = Student("Ravi", 35)

print(s1.summary())
print(s2.summary())
````

````python
class Classroom:
    def __init__(self, name):
        self.name = name
        self.students = []          # starts empty - a list that will hold Student objects

    def add_student(self, student):
        self.students.append(student)

    def class_average(self):
        if not self.students:
            return 0
        total = sum(s.marks for s in self.students)   # generator expression, like a list comprehension
        return total / len(self.students)

    def top_student(self):
        return max(self.students, key=lambda s: s.marks)

room = Classroom("CSE-3A")
room.add_student(Student("KK", 88))
room.add_student(Student("Ravi", 45))
room.add_student(Student("Anu", 92))

print("Class average:", room.class_average())
print("Top student:", room.top_student().name)
````

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):          # Dog INHERITS from Animal
    def speak(self):          # override - Dog defines its OWN version
        print(f"{self.name} says Woof!")

class GuideDog(Dog):        # inheritance can chain: GuideDog -> Dog -> Animal
    def speak(self):
        super().speak()        # super() calls the PARENT's version first
        print("I am a guide dog.")

d = Dog("Rex")
d.speak()

g = GuideDog("Buddy")
g.speak()
````

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):     # no body - subclasses MUST implement this
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width

shapes = [Circle(5), Rectangle(4, 6)]
for s in shapes:
    print(f"{s.__class__.__name__} area: {s.area():.2f}")

#shape = Shape()   # TypeError - can't instantiate an abstract class directly
````

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):             # called automatically by print(obj)
        return f"Point({self.x}, {self.y})"

    def __add__(self, other):      # called automatically by obj1 + obj2
        return Point(self.x + other.x, self.y + other.y)

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1)          # uses __str__ automatically
print(p1 + p2)      # uses __add__ automatically
````

````python
class Message:
    message_counter = 1        # class attribute, shared across every Message

    def __init__(self, sender, content):
        self.sender = sender
        self.content = content
        self.id = Message.message_counter
        Message.message_counter += 1

    def __str__(self):
        return f"({self.id}) {self.sender.username}: {self.content}"


class User:
    def __init__(self, username):
        self.username = username
        self.chatroom = None

    def join_chatroom(self, chatroom):
        if self.chatroom:
            print(f"{self.username} is already in a chatroom.")
        else:
            chatroom.add_user(self)
            self.chatroom = chatroom
            print(f"{self.username} joined {chatroom.name}")

    def send_message(self, content):
        if not self.chatroom:
            print("Join a chatroom first.")
        else:
            self.chatroom.messages.append(Message(self, content))


class Chatroom:
    def __init__(self, name):
        self.name = name
        self.users = []
        self.messages = []

    def add_user(self, user):
        self.users.append(user)

    def display_messages(self):
        for msg in self.messages:
            print(msg)      # print(msg) automatically uses Message.__str__


room = Chatroom("AIML Study Group")
alice = User("Alice")
bob = User("Bob")

alice.join_chatroom(room)
bob.join_chatroom(room)

alice.send_message("Hey Bob!")
bob.send_message("Hi Alice! Ready to study?")

room.display_messages()
````
